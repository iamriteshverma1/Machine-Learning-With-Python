Tuples are ordered collections of itwms that are immutable.
They are similar to list, but hteir immutability  makes them different

```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))

tpl = tuple()
print(type(tpl))
```

```python
nums = tuple([1,2,3,4,5,10, 9, 8, 7])
nums
```

```python
nums[1: :6]
```

```python
## Tuple Operartions -- 
nums2 = tuple([1,2,4,56,6,4,43])

nums3 = nums + nums2

print(nums3)
print(sorted(nums3))
```

```python
print(nums3 * 1)
```
